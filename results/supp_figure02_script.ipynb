```python
''' 
Script to generate Supplementary Figure 2, which shows the number of peaks detected in each frequency bin across all subjects and conditions.

'''

import pandas as pd
import os
import seaborn as sns
import matplotlib.pyplot as plt

print(os.getcwd())
datadir = os.path.abspath(os.path.join(os.path.dirname( os.getcwd() ), '.', 'data'))
print(f'Data directory: {datadir}')

# setting font size to 10
plt.rcParams.update({'font.size': 10})
plt.rcParams["font.family"] = "sans-serif"
plt.rcParams["font.sans-serif"] = ["Nimbus Sans"]

megtype = 'grad'

countsfilename = f'supp_figure02_source_data.tsv'
countsfile = os.path.join(datadir, countsfilename)


fig, axs = plt.subplots(1, 2, figsize=(7.2, 4), width_ratios=[2, 1], gridspec_kw={'wspace': 0.3, 'hspace': 0.4, 'left':0.11, 'right':.98, 'top':0.94, 'bottom':0.13})

df_counts = pd.read_csv(countsfile, sep='\t').set_index('row')

sns.barplot(data=df_counts, x='frequency_bin', y='count', color='gray', width=1, ax=axs[0])
freqs = [5, 10, 15, 20, 25, 30, 35]
axs[0].set_xticks([df_counts['frequency_bin'].tolist().index(idx) for idx in freqs])
axs[0].set_xticklabels(freqs)
axs[0].set_xlabel('Frequency Bin (Hz)')
axs[0].set_ylabel('Number of Peaks Detected')

axs[1].axis('off')

plt.savefig('Supp_Figure02.png', dpi=300)


```
